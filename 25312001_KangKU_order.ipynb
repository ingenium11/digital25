```python
# 메뉴의 이름과 가격을 변수로 지정한다.

김밥 , 떡볶이 , 라면 , 튀김 , 음료수 = 5000 , 7000 , 4000 , 8000 , 2000

# 반복문을 활용하기 위하여 리스트에 저장한다.

orderlist = ['김밥','떡볶이','라면','튀김','음료수']

# 반복문을 통해 모든 메뉴의 수량을 입력받는다.

print("☆주문내역☆")
counts = []
for menu in orderlist:
    count = int(input(f'{menu} : '))
    counts.append(count)

# 총 가격을 구한다.

price = 김밥 * counts[0] + 떡볶이 * counts[1] + 라면 * counts[2] + 튀김 * counts[3] + 음료수 * counts[4]
print('총액' , price , "원")

print("")

# if문을 이용하여 총 가격이 최소 주문가격을 충족하였는지 확인한다.

if price < 15000:
  print("최소 주문금액을 충족하지 못하였습니다")
  print('메뉴를 추가해 주세요')
else:
  print('주문이 완료되었습니다')
  print('잠시만 기다려주세요!')

print("")

print("ㅇㅇ음식점")
```

```python
import random

bus_time = {"월요일": [("08:00", "100번"), ("12:30", "200번"),("13:40", "160번"),("14:00", "230번")],
            "화요일": [("09:00", "150번"), ("14:00", "250번"),("15:30", "20번"),("15:45", "2000번")],
            "수요일": [("07:30", "120번"), ("13:45", "220번"),("14:00", "2030번"),("14:10", "2005번")],
            "목요일": [("08:30", "57번"), ("11:30", "1번"), ("12:45", "2300번"),("13:30", "2100번")],
            "금요일": [("08:35", "3번"), ("10:30", "560번"), ("12:00", "2번"), ("13:00", "28번")]}

days_order = ["월요일", "화요일", "수요일", "목요일", "금요일", "토요일", "일요일"]

while True:
    current_day = input("현재 요일을 입력하세요 (종료하려면 '종료' 입력): ").strip()
    if current_day == '종료':
        print("프로그램을 종료합니다.")
        break

    current_time = input("현재 시간을 입력하세요 (HH:MM): ").strip()

    # 입력값 검증
    try:
        hour, minute = map(int, current_time.split(':'))
        if not (0 <= hour <= 23 and 0 <= minute <= 59):
            raise ValueError
    except ValueError:
        print("잘못된 시간 형식입니다. HH:MM 형식으로 입력해주세요.")
        continue

    def find_next_bus(schedule, current_day, current_time):
        if current_day in schedule:
            today_schedule = schedule[current_day]
        else:
            return f"'{current_day}'에 해당하는 버스 시간표가 없습니다."

        found_bus = None
        for bus_time, bus_num in today_schedule:
            if current_time < bus_time:
                found_bus = (bus_time, bus_num)
                break

        if found_bus:
            return f'다음 버스는 "{found_bus[1]}" ({found_bus[0]})입니다.'

        current_idx = days_order.index(current_day)
        for i in range(1, 7):
            next_day = days_order[(current_idx + i) % 7]
            if next_day in schedule and schedule[next_day]:
                first_bus = schedule[next_day][0]
                return f'다음 요일의 첫 번째 버스는 "{first_bus[1]}" ({first_bus[0]})입니다.'

        return "남는 버스가 없습니다."

    result = find_next_bus(bus_time, current_day, current_time)
    print(result)
```

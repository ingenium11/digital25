```python
# 은행별 대출 정보 설정
banks = {
    "우리은행": {"이자율": 3.5, "최대대출": 50000000, "상한기간": 5},
    "신한은행": {"이자율": 3.8, "최대대출": 30000000, "상한기간": 7},
    "카카오뱅크": {"이자율": 4.0, "최대대출": 20000000, "상한기간": 6}
}

# 사용자 입력 받기
은행 = input("원하는 은행을 입력하세요 (우리은행, 신한은행, 카카오뱅크): ")
대출금액 = float(input("대출받을 금액을 입력하세요 (원 단위): "))

# 은행 정보가 존재하는지 확인
if 은행 not in banks:
    print("잘못된 은행 이름입니다.")
else:
    # 은행별 정보 가져오기
    은행정보 = banks[은행]
    최대대출금액 = 은행정보["최대대출"]

    # 대출 금액이 최대 금액을 초과하는지 확인
    if 대출금액 > 최대대출금액:
        print("대출 금액이 " + 은행 + "의 최대 대출 금액 " + str(최대대출금액) + "원을 초과합니다.")
    else:
        # 이자 계산
        이자 = 대출금액 * (은행정보["이자율"] / 100)
        # 총 상환 금액 계산
        총상환금액 = 대출금액 + 이자
        # 상한 기간
        상한기간 = 은행정보["상한기간"]

        # 출력
        print(은행 + "에서 대출 금액 " + str(대출금액) + "원에 대한 이자액은 " + str(이자) + "원입니다.")
        print("총 상환 금액은 " + str(총상환금액) + "원이고, 상환 기간은 최대 " + str(상한기간) + "년입니다.")
```

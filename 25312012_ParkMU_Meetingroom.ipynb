```python
meetings_list = [[1, 2], [3, 4], [0, 6], [5, 7], [8, 9], [5, 9]]

# 종료 시간을 기준으로 회의 정렬
sorted_meetings = []
while meetings_list:
    smallest = meetings_list[0]
    for meeting in meetings_list:
        if meeting[1] < smallest[1]:
            smallest = meeting
    sorted_meetings.append(smallest)
    meetings_list.remove(smallest)

# 가능한 많은 회의를 선택
selected_meetings = []
last_end_time = 0

for meeting in sorted_meetings:
    start, end = meeting
    if start >= last_end_time:  # 종료 시간이 충돌하지 않는 경우
        selected_meetings.append(meeting)
        last_end_time = end

print("선택된 회의:", selected_meetings)
```
